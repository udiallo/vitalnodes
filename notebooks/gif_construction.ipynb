```python
def construct_3_hop_neighborhood(snapshots, node_index):
    """
    Constructs a list of snapshots corresponding to the 3-hop neighborhood of the given node index.

    Parameters:
        snapshots (list): A list of NetworkX graphs representing snapshots.
        node_index (int): The index of the node for which the 3-hop neighborhood is to be constructed.

    Returns:
        list: A list of subgraphs (snapshots) corresponding to the 3-hop neighborhood of the node.
    """
    neighborhood_snapshots = []

    for snapshot in snapshots:
        if node_index in snapshot:
            # Get nodes within 3 hops of the given node
            nodes_within_3_hops = nx.ego_graph(snapshot, node_index, radius=3)
            # Add the subgraph to the result list
            neighborhood_snapshots.append(nodes_within_3_hops)

    return neighborhood_snapshots
```

```python
node_index = 31  # Change this index to select a different node
three_hop_neighborhood_snapshots = construct_3_hop_neighborhood(ABM_contact_network, node_index)
```
